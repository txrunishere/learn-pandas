```python
import pandas as pd
import numpy as np
```

```python
data = {
  'A': [1, 2, np.nan, 4, 5],
  'B': [np.nan, 2, 3, 4, 5],
  'C': [1, np.nan, np.nan, 4, 5],
  'D': [np.nan, np.nan, np.nan, 4, 5]
}

df = pd.DataFrame(data)
```

#### Finding missing data

```python
df.isna() # True for nan / None and False for other data
```

```python
df.isna().sum() # gives total missing values in each column
```

```python
df.isna().any() # if any nan / None values exists in column it returns True
```

#### Removing missing data

```python
# work on rows

df.dropna()
```

```python
df.dropna(thresh=2)
```

#### Filling the missing data

```python
df.fillna(0) # fill all none values with 0
```

```python
values = {
  "A": 0,
  "B": 10,
  "C": 20,
  "D": 30,
}
df.fillna(value=values) # fill none values as per the values dictionary
```
