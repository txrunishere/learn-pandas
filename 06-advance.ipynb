```python
import pandas as pd
import numpy as np
import re
```

```python
anime_df = pd.read_csv(r'anime.csv')
```

#### Extract episodes from title and put them in new column name episodes

```python
def extract_episodes(title: str):
  check = False
  eps = ""
  for i in title:
    if check and i == " ":
      return eps
    if check:
      eps = eps+i
    if i == '(':
      check = True
```

```python
anime_df['Episodes'] = anime_df['Title'].apply(extract_episodes).astype(int)
```

#### Make a new column for time stamp

```python
def extract_timestamp(title: str) -> str | None:
    pattern = r'([A-Za-z]{3} \d{4} - [A-Za-z]{3} \d{4})'
    match = re.search(pattern, title)
    return match.group(1) if match else None
```

```python
anime_df["Timestamp"] = anime_df['Title'].apply(extract_timestamp)
```
