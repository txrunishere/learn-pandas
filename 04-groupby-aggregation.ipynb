```python
import pandas as pd
```

```python
data = {
  'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
  'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S1', 'S2', 'S1'],
  'Sales': [100, 200, 150, 250, 340, 100, 400, 230],
  'Quantity': [10, 15, 18, 14, 8, 20, 22, 12],
  'Date': pd.date_range('2025-01-01', periods=8)
}
df = pd.DataFrame(data)
```

#### GroupBy

```python
categories = df.groupby('Category')

for category_name, values in categories:
  print(category_name)

for category_name, category_values in categories:
  print(category_values)

categories['Sales'].sum()
```

```python
stores = df.groupby('Store')

for i, v in stores:
  print(i)
  print(v)

stores['Sales'].sum()
```

```python
# Group by multiple column

mul_categories = df.groupby(["Category", "Store"])["Sales"]
mul_categories.sum()
```
