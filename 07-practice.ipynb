```python
import numpy as np
import pandas as pd
```

```python
countries_df = pd.read_csv(r'countries.csv')
```

##### Which country has the highest population

```python
countries_df[countries_df['population'] == countries_df['population'].max()]
```

##### What is the capital of the country with highest population

```python
countries_df[countries_df['population'] == countries_df['population'].max()]['capital_city']
```

##### Which country has the least population

```python
countries_df[countries_df['population'] == countries_df['population'].min()]
```

##### Top 5 countries with highest democratic score

```python
countries_df.sort_values(by='democracy_score', ascending=False).head()
```

##### How many total regions are there

```python
countries_df['region'].value_counts()
```

##### How many countries are lie there in Eastern Europe region

```python
countries_df[countries_df['region'] == 'Eastern Europe']['country']
```

##### Who is the political leader of the 2nd highest populated country

```python
countries_df.sort_values(by='population', ascending=False)[['country', 'political_leader']].iloc[1]
```

```python
countries_df[countries_df['population'] == countries_df['population'].nlargest(2).iloc[1]]['political_leader']
```

##### How many countries are there whoes political leaders are Unknown

```python
countries_df.loc[countries_df['political_leader'].isna()]['country']
```

##### How many countries have Republic in their long / full name

```python
def check_country(country: str):
  return True if country.find("Republic") > -1 else False
```

```python
countries_df[countries_df['country_long'].apply(check_country)]['country']
```

```python
countries_df[countries_df['country_long'].str.contains('Republic', na=False)]
```

##### Which country in african region has highest population

```python
africa_df = countries_df[countries_df['continent'] == 'Africa']
africa_df[africa_df['population'] == africa_df['population'].max()]
```
