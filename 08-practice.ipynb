```python
# Create a column:

# "High" if salary > 55000
# "Medium" if 45000–55000
# "Low" otherwise
```

```python
def salary_level_categorize(x: int):
  if x > 55000:
    return "high"
  elif 45000 <= x <= 55000:
    return "medium"
  else:
    return "low"
```

```python
employees_df['salary_level'] = employees_df['salary'].apply(salary_level_categorize)
employees_df
```

```python
# Find employees:

# from IT department
# age < 30
# salary > 50000
```

```python
it_df = employees_df[(employees_df['department'] == "IT") & (employees_df['age'] > 30)]
it_df[it_df['salary'] > 50000]
```
