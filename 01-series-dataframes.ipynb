## DataFrames

#### Creating DataFrames

```python
restaurants_sale = [
  [42000, 45000, 47000, 49000, 50000],
  [60000, 65000, 70000, 72000, 75000],
  [35000, 38000, 39000, 41000, 42000],
  [50000, 55000, 60000, 64000, 68000] 
]

sales_columns = [2021, 2022, 2023, 2024, 2025]
restaurants = ["Spice Hub", "Food Palace", "Green Leaf", "Urban Bites"]

restaurants_df = pd.DataFrame(restaurants_sale, columns=sales_columns, index=restaurants)
```

#### Selection and Indexing of Columns

```python
user_df["Name"] # selects single column and return a series
```

```python
user_df[['Name', 'Age']] # select multiple columns
```

#### Creating a new column

```python
user_df['Gender'] = ["male", 'male', 'female', 'female']
user_df
```

#### Removing column

```python
user_df.drop('Gender', axis=1, inplace=True) # axis = 1 for rows
```

```python
user_df.drop(0, axis=0) # axis = 0 for columns
```

#### Selecting Rows

```python
# loc for select row using its name
restaurants_df.loc['Food Palace'] # select single row
restaurants_df.loc[['Green Leaf', 'Food Palace']] # select multiple rows

# iloc for select rows using their index values instead of names
restaurants_df.iloc[0] # select single row
restaurants_df.iloc[[0, 3]] # select multiple rows
```

#### Selecting subsets of rows and columns

```python
user_df.iloc[[0,1]][['City', 'Salary']]
```

```python
restaurants_df.iloc[[1, 2]][[2023, 2024, 2025]]
```
