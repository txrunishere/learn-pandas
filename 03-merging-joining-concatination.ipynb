```python
import pandas as pd
import numpy as np
```

```python
employees = pd.DataFrame({
  "employee_id": [101, 102, 103, 104],
  "name": ["John", "Bob", "Alina", "Anna"],
  "department": ["IT", "HR", "Finance", "Marketing"]
})

salaries = pd.DataFrame({
  "employee_id": [101, 102, 107, 108],
  "salary": [50000, 60000, 55000, 52000],
  "bonus": [5000, 6000, 5500, 5200]
})
```

#### Merging of dataframes

```python
pd.merge(employees, salaries)
```

```python
pd.merge(employees, salaries, on='employee_id') # merge on the basis of employee_id
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner') # on the basis of row if two df contains same value as employee_id merge take place
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer') # on the basis of column
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
```

#### Concatination of dataframes

```python
df1 = pd.DataFrame({
  'A': ['A0', 'A1', 'A2'],
  'B': ['B0', 'B1', 'B2'],
  'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
  'A': ['A3', 'A4', 'A5'],
  'B': ['B3', 'B4', 'B5'],
  'C': ['C3', 'C4', 'C5']
})
```

```python
pd.concat([df1, df2], axis=0) # concatination occours on column level
```

```python
pd.concat([df1, df2], axis=1) # concatination occours on rows level
```

#### Joining of dataframes

```python
names = pd.DataFrame({
  "name": ["Alice", "Bob", "Charlie"]
}, index=[1, 2, 3])

scores = pd.DataFrame({
  "score": [85, 90, 75]
}, index=[2, 3, 4])
```

```python
names.join(scores, how='inner')
```
